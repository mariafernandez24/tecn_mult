## 0. El cuaderno

Un `.ipynb` es un cuaderno de Jupyter, y no es un programa. Tampoco es un fichero de Python: un
`.py` es texto que se ejecuta entero y de una vez, y esto es un fichero en formato JSON que guarda
una lista de celdas, el tipo de cada una y también lo que cada una imprimió la última vez que se
ejecutó, imágenes incluidas. Por eso un cuaderno se abre con los resultados del día anterior ya
dentro, sin haber ejecutado nada.

Las celdas son de dos tipos. Las de texto están escritas en markdown, un formato en el que una
almohadilla al principio de la línea hace un título, los asteriscos ponen **negrita** y las
comillas invertidas marcan un `nombre de fichero`; la ventana lo enseña compuesto, que es como se
está leyendo esta misma celda. Las de código son texto de Python, y son las que traen el trabajo
del curso.

El código no lo ejecuta el fichero. Al ejecutar una celda de código, con *Mayús+Intro* o con el
triángulo de su izquierda, la ventana manda ese texto a un programa aparte llamado **kernel**, que
lo ejecuta y devuelve el resultado; lo que aparece debajo de la celda es esa respuesta, y es lo
que acaba guardado en el JSON. El kernel arranca al abrir el cuaderno y se queda esperando la
celda siguiente. Cuál de los Python de la máquina es el que está detrás, y cómo se elige, es el
1.5.

Tres convenios que seguiremos durante todo el curso:

1. Un cuaderno se ejecuta de arriba abajo.
2. Las importaciones van todas en la primera celda.
3. El cuaderno es para explorar y explicar.

## 1. Python, librerías y entornos

Python es un lenguaje interpretado. Un código en Python es un fichero `.py`, que no es más que
texto. El procesador no entiende ese texto, así que alguien tiene que traducirlo: un programa
llamado `python` (en Windows `python.exe`), que lo lee y lo va ejecutando sobre la marcha cada
vez que lo lanzamos.

Tener Python instalado es, por tanto, tener una copia de ese programa en una carpeta. Podemos
tener varias, de versiones distintas y con librerías distintas. De hecho, es bastante normal.

Nota: que un archivo de Python se ejecute sobre la marcha no significa que vaya línea por línea
desde el principio. Primero lee el fichero entero y comprueba que está bien escrito: si hay un
fallo de sintaxis en la última línea, no se ejecuta ni la primera. Superado ese filtro ejecuta
de arriba abajo, y ahí los fallos aparecen solo si el programa llega a pasar por ellos.

### 1.1 Instalar Python

Hay dos formas de conseguir el programa `python.exe`:

La directa es descargarlo de `python.org`. Deja un Python instalado y listo, y para empezar va
bien. El problema aparece con el tiempo: es **un solo** Python para todo lo que hagamos en esa

forma de contentar a los dos.

La otra es instalar un gestor, y el que usaremos aquí es **conda**, en su versión ligera
Miniconda. No instala un `python.exe`: instala un programa capaz de crear tantos como queramos,
cada uno por su cuenta. Eso es lo que veremos en el apartado siguiente.

En el laboratorio ya está puesto y no hay que instalar nada. En vuestro PC sí tendréis que
descargar Miniconda de su página oficial, aceptar las opciones por defecto y, al terminar,
la orden `conda` quedará disponible en el terminal.

### 1.2 Entornos

Un entorno es una carpeta con su propio intérprete y sus propias librerías, independiente del
resto. Es, literalmente, otra de esas copias de Python de las que hablábamos arriba, con lo que
decidamos poner dentro.

**¿Dónde se escribe todo esto?** Los comandos `conda` van en un **terminal**, no en celdas del
cuaderno. En Windows, concretamente en el **Anaconda Prompt**, que se busca por su nombre en el
menú de inicio; el `cmd` o el PowerShell normales no conocen la orden `conda`. En Linux y macOS
vale el terminal de siempre.

Tampoco sirve ponerlos con `!` en una celda: cada `!` abre un terminal nuevo que muere al acabar
la línea, así que un `conda activate` escrito así no dura nada.

La orden para crear un entorno es:

```bash
conda create -n NOMBRE -c CANAL PAQUETES...
```

Cada trozo de esa línea dice una cosa:

- `conda create` es la orden que crea el entorno nuevo.
- `-n NOMBRE` le pone el nombre, que es con el que entraremos luego.
- `-c CANAL` indica de dónde se bajan los paquetes.
- `PAQUETES...` es la lista de lo que queremos dentro, separada por espacios.

El canal es el repositorio de descarga. Usaremos `conda-forge`, el comunitario, más completo y
más al día que el que conda trae por defecto, y el que tiene las versiones de `opencv` y
`ffmpeg` que necesitamos en este curso.

### 1.3 Módulos, paquetes y librerías

Veremos estas tres palabras usadas casi como sinónimos, y conviene saber qué hay detrás de cada
una.

Un **módulo** es un fichero `.py`. Entre los archivos de la sesión, junto a este cuaderno, hay un
módulo `saludos.py`. Conviene abrirlo y mirarlo antes de seguir: dentro solo hay dos funciones. Se
importa por su nombre sin la extensión, y sus funciones se usan poniendo delante el nombre del
módulo.

````python
import saludos                    # carga el fichero saludos.py

saludos.saludar("Ana")            # nombre del módulo, punto, nombre de la función
````

Un **paquete** es una carpeta de módulos. Entre los archivos de la sesión hay dos carpetas, es
decir, dos paquetes: `despedidas_sin_init` y `despedidas_con_init`. Cada una lleva dentro los
mismos dos módulos, `formal` y `corta`, con el mismo código. Los dos definen una función `adios`,
y no chocan porque cada una vive en su módulo.

Empezamos por el paquete `despedidas_sin_init`. Importamos la carpeta e intentamos usar las
funciones de los módulos que tiene dentro.

````python
import despedidas_sin_init
print("Importado sin errores:", despedidas_sin_init.__name__)
despedidas_sin_init.formal.adios("Ana")     # ← esto va a fallar
````

`AttributeError`. El `import` no ha dado ningún problema, pero lo que ha traído no incluye los
módulos: para Python, `despedidas_sin_init` es la carpeta, y `formal` es otra cosa que todavía
no ha cargado. Un paquete grande tendría que leerse entero cada vez, así que solo se carga lo
que se pida; y aquí no hemos pedido nada concreto, así que no se ha cargado nada.

Si el paquete llega vacío, podríamos pensar en sacar directamente de él la función que nos
interesa:

````python
from despedidas_sin_init import adios       # ← esto también va a fallar
````

`ImportError`. En el paquete no hay ninguna función `adios`: las funciones viven dentro de los
módulos, y el paquete solo es la carpeta que los contiene.

Hay que pasar por el módulo.

````python
from despedidas_sin_init import formal
formal.adios("Ana")
````

Ahora sí funciona. Hemos nombrado el módulo, Python lo ha cargado y tenemos su función.

Conviene volver a ejecutar la primera celda, la que fallaba con `AttributeError`. Ahora debería
funcionar, sin haberle cambiado nada: al cargar `formal`, Python lo ha dejado enganchado al
paquete, así que `despedidas_sin_init.formal` ya existe.

El código es el mismo y el resultado es otro. Lo que ha cambiado está en el kernel, no en la
celda.

Pasamos ahora al otro paquete, `despedidas_con_init`. Dentro tiene los mismos dos módulos, con
el mismo código, más un fichero llamado `__init__.py`. Ese fichero es especial: Python lo
ejecuta al importar la carpeta, y ahí se decide qué queda a la vista. Conviene abrirlo y leerlo
antes de seguir.

Vamos a repetir lo que hicimos antes, a ver qué cambia. Importamos la carpeta sola, sin nombrar
ningún módulo, y llamamos a las funciones de dentro. Conviene leer los comentarios del código y
predecir qué sale en cada línea antes de ejecutar.

````python
import despedidas_con_init

despedidas_con_init.formal.adios("Ana")   # antes falló: no habíamos nombrado formal
despedidas_con_init.corta.adios("Ana")    # lo mismo con corta
despedidas_con_init.adios("Ana")          # aquí no hemos dicho cuál de los dos queremos
````

La tercera línea, `despedidas_con_init.adios("Ana")`, no dice si queremos la despedida formal o la
corta, y sin embargo sale una.

Ese `__init__.py` se ha ejecutado al importar, y por eso ha aparecido su mensaje. Además de
dejar los módulos a la vista, ha subido una función al nivel del paquete, así que ahora sí se
puede sacar directamente de él.

````python
from despedidas_con_init import adios
adios("Ana")
````

Esto es justo lo que fallaba en el otro paquete. No ha cambiado el código de los módulos: ha
cambiado lo que el `__init__.py` pone a la vista.

Ese fichero decide qué se ve del paquete, pero no puede enseñar lo que no está dentro de la
carpeta.

````python
from despedidas_con_init import saludar     # ← esto va a fallar
````

`ImportError`. `saludar` está en `saludos.py`, que no tiene nada que ver con este paquete.

El paquete es el mismo; lo único que cambia es cómo lo llamamos. Por eso aparecen tanto
`import numpy as np` y `import matplotlib.pyplot as plt`: son nombres largos que se escriben
muchas veces.

Una **librería** no es un término técnico de Python, sino la palabra corriente para «código de
otros que reutilizamos». Puede ser un módulo o un paquete; al hablar nadie distingue, y en este
cuaderno tampoco. Es la palabra que usaremos el resto del curso.

### 1.4 Las librerías

Unas pocas vienen con la propia instalación de Python: `os`, `json`, `pathlib`. Forman la
librería estándar y siempre están. Las demás hay que instalarlas, y en este cuaderno usaremos
`numpy` y `matplotlib` entre otras.

Casi ninguna de esas librerías está escrita en Python, y el motivo es la velocidad. Un programa
en C se traduce entero a instrucciones del procesador antes de ejecutarse, de una vez; en Python
lo que se ejecuta es un intérprete que va decidiendo sobre la marcha qué hacer con cada
instrucción. Esa decisión se paga en cada vuelta de un bucle, así que recorrer un millón de
píxeles sale carísimo: lo que en C tarda un milisegundo aquí puede tardar cientos.

Por eso las librerías de cálculo llevan dentro código C ya compilado. Cuando llamamos a NumPy,
la cuenta la hace ese C; nuestro Python solo le pasa los datos y recoge el resultado. En el
apartado 4 mediremos la diferencia.

### 1.5 El kernel

Un entorno es una carpeta con su propio `python.exe`, y en la misma máquina puede haber varios.
Para ejecutar una celda hace falta uno, así que la primera pregunta de cualquier cuaderno es cuál
le ha tocado.

Al abrir un `.ipynb` no arranca un programa, sino dos. Uno es la ventana que vemos, VS Code o
Jupyter, que enseña las celdas y guarda el fichero; de Python no sabe nada. El otro es el
**kernel**: un proceso que arranca uno de esos `python.exe` y se queda esperando. Al ejecutar una
celda, la ventana le manda el texto, el kernel lo ejecuta y devuelve el resultado, que es lo que
aparece debajo y lo que acaba guardado en el JSON.

Conviene separar tres palabras que se usan como si fueran la misma:

- El **entorno** es una carpeta con un `python.exe` dentro y sus librerías al lado.
- El **intérprete** es ese `python.exe`, un fichero que está en el disco.
- El **kernel** es ese mismo fichero ejecutándose ahora, atendiendo a las celdas de este
  cuaderno.

No son tres programas: son el mismo visto con más o menos detalle. Si borramos el entorno, el
cuaderno se queda sin kernel posible; si matamos el kernel, el entorno sigue intacto en el disco.

Que el entorno exista no implica que el cuaderno lo use. VS Code elige por su cuenta y a menudo
elige otro, con lo que el kernel arranca sobre el intérprete equivocado. El síntoma es un
`ModuleNotFoundError` de una librería que sabemos instalada: lo está, pero en la carpeta de otro
entorno.

La celda siguiente le pregunta al kernel qué intérprete está usando, en qué entorno vive y si
encuentra las librerías del curso. No las carga: solo comprueba que sabría dónde buscarlas. La
ruta depende de cada máquina.

````python
import sys, os, importlib.util

print("Python     :", sys.version.split()[0])
print("Intérprete :", sys.executable)
print("Entorno    :", os.environ.get("CONDA_DEFAULT_ENV", "(ninguno)"))
print()
for nombre in ["numpy", "matplotlib"]:
    hay = importlib.util.find_spec(nombre) is not None
    print(f"{'OK ' if hay else 'NO '} {nombre}")
````

Esa ruta es el `python.exe` que está ejecutando estas celdas, y `CONDA_DEFAULT_ENV` es el entorno
donde vive. Si alguna librería sale como `NO`, o no está instalada en ese entorno o el kernel es
de otro. Se cambia en el selector de kernel, arriba a la derecha: lo que hace esa orden es matar
el proceso actual y arrancar otro sobre el intérprete elegido. Conviene mirar esta ruta antes de
reinstalar nada.

El kernel no termina al acabar la celda. Se queda esperando la siguiente, y con él sigue vivo todo
lo que la celda haya dejado creado. De eso va el apartado 2.

Lo que cambia es quién recoge el texto antes de dárselo. En la primera ventana hablamos con

propias órdenes y le pasa a `python.exe` solo lo que es código Python.

Ese es el kernel de este cuaderno de Jupyter: el `python.exe` cuya ruta salió en el 1.5, con
IPython por delante. Los errores del 1.3 salieron con el mismo bloque de flechas que la división
por cero de la segunda ventana, y ahora se sabe quién los dibujaba.

El `%whos` ha enseñado además lo que quedó en memoria al ejecutar la celda del 1.5: las variables
`nombre` y `hay` del bucle, y los módulos que importó.

````python
2 + 2
3 + 3
````

En la ventana de `python` salieron el `4` y el `6`. Aquí solo sale el `6`.

Lo que cambia es qué cuenta como un turno de la conversación. En el `>>>` el turno es la línea, y
se contesta a cada una. En el cuaderno el turno es la celda entera, así que se contesta una sola
vez, con el valor de la última expresión: el 4 se calculó y se tiró. Para verlo hace falta un
`print`.

Un fichero no contesta nunca. Un `.py` que contenga `2 + 2` no saca nada por pantalla, porque no
hay conversación: es un programa que se ejecuta y termina. Quien venga de MATLAB llega esperando
lo contrario, porque un `.m` sí va enseñando lo que calcula.

### 1.7 Las órdenes de IPython

`%whos` y `len?` son dos de las órdenes que IPython añade por su cuenta. Hay muchas más, y todas
sirven para cosas que se hacen mientras se trabaja y que no forman parte del programa: cronometrar
un trozo de código, listar las variables vivas, borrarlas todas. Con `%lsmagic` sale la lista
completa.

Hay dos clases, y se distinguen por el número de símbolos. Las *line magics* llevan un `%` y
afectan solo a esa línea. Las *cell magics* llevan `%%`, tienen que ir en la primera línea de la
celda y se quedan con la celda entera.

`%%timeit` es de las segundas: ejecuta el código muchas veces y da una estadística del tiempo.

imprime la media y la desviación típica. Esa desviación sale de lo que el ordenador esté haciendo
por su cuenta, que siempre suma tiempo y nunca lo resta.

````python
%%timeit -n 3 -r 3
sum(range(200_000))
````

## 3. Sintaxis de Python

Quien venga de C, de Java o de MATLAB se va a encontrar con cinco diferencias desde la primera
línea que escriba:

- Los bloques no se delimitan con llaves ni con un `end`, sino con la indentación (apartado 3.1).
- El tipo no se declara en la variable: lo lleva el valor.
- El primer elemento de una secuencia es `v[0]`, no `v(1)` como en MATLAB.
- Para imprimir con formato, en lugar de `printf("%d\n", x)` se escribe `print(f"{x}")`, que es
  lo que se llama una *f-string*.
- Un comentario empieza por `#`. No existen `//` ni `/* */`.

### 3.1 Los dos puntos y la indentación

En Python los bloques no se delimitan con llaves ni con un `end`: los delimita la indentación.
Un bloque se abre con dos puntos al final de la línea y lo forman las líneas siguientes que
estén escritas más a la derecha. El bloque termina en cuanto la indentación vuelve al nivel
anterior.

Cada nivel son cuatro espacios. La tecla *Tab* vale para sangrar, porque VS Code la convierte
en espacios, y en la barra de abajo se lee «Espacios: 4». Lo que da guerra es el código pegado
de fuera: si viene sangrado con tabuladores de verdad, el fichero mezcla las dos formas y
Python da `TabError` sin que en pantalla se vea nada torcido.

Cuántos espacios lleva cada nivel es una convención, y por eso se puede elegir. En qué nivel está
cada línea, no: mover una línea un nivel a la izquierda o a la derecha cambia lo que hace el
programa, y sin dar error. La indentación es sintaxis, no presentación.

### 3.2 Ejemplo resuelto: cambia el significado sin dar error

Las dos funciones se diferencian en dónde está el `return`. Ninguna da error.

````python
def sumar_positivos_mal(numeros):
    total = 0
    for n in numeros:
        if n > 0:
            total += n
        return total          # DENTRO del for: sale en la primera vuelta


def sumar_positivos_bien(numeros):
    total = 0
    for n in numeros:
        if n > 0:
            total += n
    return total              # FUERA del for: sale al terminar


datos = [5, -3, 10, 2]
print("Esperado :", 17)
print("Mal      :", sumar_positivos_mal(datos), "  <- ninguna excepción")
print("Bien     :", sumar_positivos_bien(datos))
````

Las dos funciones se diferencian en cuatro espacios, ninguna da error y una de las dos devuelve un
resultado equivocado.

### 3.4 Bucles

Python tiene dos bucles, `for` y `while`. El `for` no es el de C: no lleva contador, ni condición,
ni incremento, sino que recorre el contenido de una colección y da en cada vuelta uno de sus
elementos. Debajo van las tres maneras de recorrer que se usan durante el curso, la forma de
contar vueltas cuando no hay nada que recorrer, y el `while`.

#### Recorrer una colección

La forma normal del `for` es nombrar la colección y una variable que recibe un elemento en cada
vuelta. Escribir `for i in range(len(librerias))` para luego usar `librerias[i]` es traducir el
bucle de C, y casi nunca hace falta.

````python
librerias = ["numpy", "matplotlib", "pathlib"]

for nombre in librerias:          # la variable recibe el elemento, no su posición
    print("cargando", nombre)
````

#### Contar vueltas con `range`

Cuando no hay colección que recorrer sino un número de vueltas que dar, se cuenta con `range`.
**`range` excluye el valor final.** `range(3)` da 0, 1 y 2; en MATLAB `1:3` sí incluye el 3.
Admite además un principio y un paso.

````python
for i in range(3):                # 0, 1, 2: el final se excluye
    print("vuelta", i)

print(list(range(2, 10)))         # del 2 al 9
print(list(range(0, 10, 2)))      # los pares: principio, final y paso
````

#### La posición con `enumerate`, dos colecciones con `zip`

Quedan los dos casos intermedios. Si además del elemento hace falta saber por qué posición va,
`enumerate` devuelve las dos cosas a la vez; y si hay que recorrer dos colecciones en paralelo,
emparejando el primero con el primero, `zip` las junta en pares. Los dos dan dos cosas en cada
vuelta, y por eso sus `for` llevan dos variables. Con ellos no hay que llevar el índice a mano,
que es de donde salen los errores de índice.

Si las dos listas tuvieran distinta longitud, el bucle pararía al acabarse la más corta: sin
error y sin aviso, con lo que el resultado sale con menos vueltas de las debidas y con aspecto de
correcto. Ocurre en cuanto las dos listas se construyen por separado y una pierde un elemento por
el camino. Para que el descuadre salte en vez de pasar inadvertido, `zip(a, b, strict=True)`
levanta un `ValueError` cuando las longitudes no coinciden.

````python
for i, nombre in enumerate(librerias):    # el índice y el elemento
    print(i, nombre)

anchos, altos = [854, 1280], [480, 720]
for a, b in zip(anchos, altos):           # un elemento de cada lista
    print(f"{a}x{b}")

altos.append(1080)                        # ahora altos tiene un elemento más
print(list(zip(anchos, altos)))           # se pierde el 1080 y no avisa

try:
    print(list(zip(anchos, altos, strict=True)))
except ValueError as e:
    print("ValueError:", e)
````

#### El `while`

El `while` es el único que lleva condición, y sirve para cuando no se sabe cuántas vueltas van a
ser: leer imágenes de un vídeo hasta que se acabe, por ejemplo. La condición se comprueba antes de
cada vuelta, así que algo del bloque tiene que acabar haciéndola falsa. No existe `i++`: se
escribe `n += 1`. Dentro de cualquiera de los dos bucles, `break` lo corta y `continue` salta a la
vuelta siguiente.

````python
n = 3
while n > 0:                      # cuando no se sabe cuántas vueltas
    n -= 1                        # no existe n--
print("fin del while, n =", n)

for nombre in librerias:
    if nombre == "matplotlib":
        continue                  # salta a la vuelta siguiente
    if nombre == "pathlib":
        break                     # corta el bucle
    print(nombre)
````

### 3.5 Condiciones

Una condición se escribe con `if`, y las ramas alternativas con `elif` y `else`. Los tres llevan
dos puntos al final y su bloque indentado debajo. En Python no existe el `switch` de C.

Tres diferencias con C que dan errores de sintaxis: se escribe `and`, `or` y `not` en vez de
`&&`, `||` y `!`; `True` y `False` con mayúscula; y el valor nulo es `None`, no `NULL`.

````python
valor = 300

if valor < 0:
    print("por debajo de 0")
elif valor > 255:
    print("por encima de 255")
else:
    print("dentro del rango")

print(0 <= 128 <= 255)                  # comparación encadenada, sin and
print((valor > 0) and not (valor < 256))
print("numpy" in librerias)             # pertenencia a una colección
print(bool([]), bool([1]), bool(""), bool("a"))   # vacío es falso
````

`0 <= x <= 255` se escribe tal cual, y sirve para comprobar rangos de imagen.

Una lista o cadena vacía cuenta como falsa: `if lista:` pregunta si tiene elementos.

### 3.7 Estructuras de datos

Hasta ahora cada variable guardaba un solo valor. Para guardar varios hay tres estructuras, y lo
que las distingue es cómo se saca después un dato de dentro: por su posición, por un nombre, o
todos a la vez. Debajo va cada una con los datos de vídeo que se usan durante el curso, y al final
una forma abreviada de construir listas.

#### Listas

En una **lista** los elementos están en orden y se piden por su posición, contando desde el 0. El
índice `-1` es el último, y evita tener que escribir `len(lista) - 1`. Se puede modificar una vez
creada, así que sirve cuando no se sabe de antemano cuántos elementos va a haber ni cuáles: es lo
que se usa para ir acumulando resultados dentro de un bucle.

````python
extensiones = ["mp4", "avi", "mkv"]                 # ordenada y modificable
print(extensiones[0], extensiones[-1], len(extensiones))   # el -1 es el último
extensiones.append("mov")                           # añadir al final
print(extensiones)
````

#### Diccionarios

Un **diccionario** no tiene posiciones. Guarda pares de clave y valor, y dentro se busca por la
clave: por `"ancho"`, no por el 0. Sirve cuando cada dato tiene nombre propio y el orden en que se
escribieron no significa nada. Pedir una clave que no está da `KeyError`, y por eso existe `.get`,
que devuelve lo que se le indique en vez de fallar. Para recorrerlo está `.items()`, que da dos
cosas en cada vuelta, la clave y el valor, y por eso el `for` lleva dos variables.

````python
dimensiones = {"ancho": 1920, "alto": 1080}         # pares de clave y valor
print(dimensiones["ancho"])
print(dimensiones.get("fps", "no consta"))          # get no falla si la clave no está

for clave, valor in dimensiones.items():            # la clave y el valor
    print(" ", clave, "=", valor)
````

#### Tuplas

Una **tupla** se escribe entre paréntesis y no admite cambios una vez creada. Agrupa valores que
van juntos y se leen a la vez, como el ancho y el alto de un formato de vídeo: `(1920, 1080)` es
un dato solo escrito con dos números, y el orden forma parte de lo que significa. No suele
recorrerse, sino desempaquetarse: se ponen a la izquierda tantos nombres como valores trae y cada
uno recibe el suyo. Intentar modificar un elemento da `TypeError`, y esa es toda la diferencia con
la lista.

````python
forma = (1080, 1920, 3)                             # alto, ancho y canales de una imagen
alto, ancho, canales = forma                        # desempaquetado: un nombre por número
print(f"{ancho}x{alto}, {canales} canales")

try:
    forma[0] = 720                                  # una tupla no admite cambios
except TypeError as e:
    print("TypeError:", e)
````

#### Comprensiones de lista

Construir una lista recorriendo otra y quedándose con algunos elementos lleva tres líneas: crear
la lista vacía, el `for` y el `append` dentro de un `if`. La **comprensión de lista** escribe eso
mismo en una, entre corchetes y en el orden en que se lee: qué se guarda, de dónde sale y qué
condición cumple. La primera celda lo hace con el bucle y la segunda con la comprensión, y las dos
imprimen la misma lista.

````python
grandes = []
for t in [480, 720, 1080, 2160]:
    if t >= 1080:
        grandes.append(f"{t}p")
print(grandes)
````

````python
grandes = [f"{t}p" for t in [480, 720, 1080, 2160] if t >= 1080]
print(grandes)                                      # comprensión de lista
````

### 3.8 Funciones

Una función se declara con `def`, el nombre, los argumentos entre paréntesis y dos puntos, y el
cuerpo va indentado debajo. Los argumentos admiten valor por defecto y se pasan por posición o por
nombre. Los que van por posición tienen que respetar el orden del `def`; los que van por nombre no,
porque el nombre ya dice a qué argumento corresponde cada valor.

Las de este curso llevan además **anotaciones de tipo**, que dicen de qué tipo se espera cada
argumento y qué devuelve, y **docstring**, que es una cadena de texto puesta como primera línea
del cuerpo y que explica qué hace la función. Es lo que devuelve `help(ajustar_brillo)` y lo que
enseña el editor al pasar el ratón por encima del nombre; el `len?` del 1.6 mostraba el de `len`.
Ninguna de las dos cosas es obligatoria.

````python
def ajustar_brillo(valor: int, factor: float = 2.0) -> int:
    """Multiplica el brillo de un píxel y lo recorta a un máximo de 255."""
    resultado = int(valor * factor)
    if resultado > 255:
        return 255
    return resultado


print(ajustar_brillo(100))                        # usa el factor por defecto
print(ajustar_brillo(200))                        # se recorta a 255
print(ajustar_brillo(50, 3))                      # los dos por posición
print(ajustar_brillo(50, factor=3))               # el segundo por nombre
print(ajustar_brillo(factor=3, valor=50))         # los dos por nombre y en otro orden


def mostrar_brillo(valor: int) -> None:
    """Imprime el brillo de un píxel y no devuelve nada."""
    print("brillo:", valor)

devuelto = mostrar_brillo(100)
print(devuelto)                                   # None: la función no lleva return
````

`mostrar_brillo` imprime el valor, pero no lo devuelve: sin `return`, la llamada vale `None`, y
eso es lo que ha guardado `devuelto`. Si una variable vale `None` sin motivo, conviene mirar si
falta el `return`.

### 3.10 Rutas

Una ruta es el nombre completo de un fichero, con las carpetas que hay que atravesar para llegar

`pathlib`, de la librería estándar, resuelve eso con un objeto `Path`. Se construye a partir de un

````python
from pathlib import Path

ruta = Path("despedidas_con_init") / "formal.py"   # carpeta / fichero

print("Ruta completa :", ruta)
print("Nombre        :", ruta.name)      # formal.py
print("Sin extensión :", ruta.stem)      # formal
print("Extensión     :", ruta.suffix)    # .py
print("Carpeta       :", ruta.parent)    # despedidas_con_init
print("¿Existe?      :", ruta.exists())  # True
````

`ruta` es un objeto, no una cadena de texto: por eso tiene `.stem`, `.suffix` y `.exists()`, que

## 4. numpy: la imagen como matriz

Una imagen en memoria es una matriz de números. En numpy tiene tres dimensiones: alto, ancho y
canales de color. Cada elemento es la intensidad de un color en un píxel, normalmente un entero
de 0 a 255. Un vídeo es una secuencia de esas matrices.

### 4.1 Ejemplo resuelto: el orden de las dimensiones

El orden de las dimensiones no es evidente, así que conviene verlo escrito.

````python
import numpy as np

bloque = np.array([
    [[255, 0, 0], [0, 255, 0]],       # fila 0: rojo,  verde
    [[0, 0, 255], [255, 255, 0]],     # fila 1: azul,  amarillo
], dtype=np.uint8)

print("shape :", bloque.shape)                     # (alto, ancho, canales)
print("dtype :", bloque.dtype)                     # uint8: enteros de 0 a 255
print("píxel (0,0)      :", bloque[0, 0])          # los tres canales de un píxel
print("canal rojo       :\n", bloque[:, :, 0])     # matriz 2x2
````

- `shape` va en orden **(alto, ancho, canales)**: primero la fila. Es al revés que «1920×1080»
  y da muchos errores.
- `bloque[:, :, 0]` coge un canal entero. Los dos puntos son «todo este eje».
- El tipo `uint8` guarda enteros de 0 a 255, es decir un byte por canal.

### 4.2 Ejemplo resuelto: aritmética con uint8

`uint8` va de 0 a 255. Al salirse no se satura: da la vuelta. Afecta a cualquier resta entre
imágenes.

````python
clara  = np.array([[100, 200]], dtype=np.uint8)
oscura = np.array([[120, 180]], dtype=np.uint8)

print("Resta directa  :", clara - oscura, "  ← el -20 se ha convertido en 236")
print("Resta correcta :", clara.astype(np.int16) - oscura.astype(np.int16))
````

`uint8` es el formato en que se guardan y se muestran las imágenes, pero no sirve para hacer
cuentas con ellas: cualquier resta puede dar negativo y cualquier suma puede pasar de 255.
`int16` va de −32768 a 32767, así que ahí cabe cualquier suma o resta de dos bytes; `float32`
admite además decimales, para multiplicar por coeficientes. Por eso se convierte antes de operar
y se vuelve a `uint8` al final, cuando el resultado ya está entre 0 y 255 y va a guardarse o
dibujarse.

### 4.3 Verificación

Comparar dos imágenes píxel a píxel es restarlas, pero la resta tiene un signo que aquí no
interesa: un píxel que sube 20 y otro que baja 20 han cambiado lo mismo. Por eso se toma el valor
absoluto del resultado, que en NumPy es `np.abs` y se aplica al array entero de una vez.

La función siguiente hace justo eso. Leída, parece correcta: dice en su primera línea lo que hace,
hace lo que dice su nombre y al ejecutarla no da ningún error. Es lo que escribe cualquiera que no
haya pasado por el 4.2, incluido un asistente de programación al que se le pida.

```python
def cuanto_ha_cambiado(img1, img2):
    """Devuelve cuánto ha cambiado cada píxel entre dos imágenes."""
    return np.abs(img1 - img2)
```

### 4.4 Ejemplo resuelto: vectorización

Recorrer píxeles con bucles es lo natural viniendo de C, pero cada iteración pasa por el
intérprete. Lo correcto es operar sobre el array entero y que numpy lo resuelva en C.

Las dos funciones hacen la misma cuenta: convierten la imagen en color a una en gris, combinando
los tres canales de cada píxel con tres coeficientes fijos. De dónde salen esos tres números no
importa hoy.

````python
import time

rng = np.random.default_rng(0)
img_aleatoria = rng.integers(0, 256, size=(400, 400, 3), dtype=np.uint8)

def a_gris_con_bucles(im):
    alto, ancho, _ = im.shape
    salida = np.empty((alto, ancho), dtype=np.float32)
    for y in range(alto):
        for x in range(ancho):
            r, g, b = im[y, x]
            salida[y, x] = 0.299 * r + 0.587 * g + 0.114 * b
    return salida

def a_gris(im):
    return im[:, :, 0] * 0.299 + im[:, :, 1] * 0.587 + im[:, :, 2] * 0.114

t0 = time.perf_counter(); a = a_gris_con_bucles(img_aleatoria);   t_bucle = time.perf_counter() - t0
t0 = time.perf_counter(); b = a_gris(img_aleatoria);  t_vect  = time.perf_counter() - t0

print(f"Con bucles  : {t_bucle*1000:8.1f} ms")
print(f"Vectorizado : {t_vect*1000:8.1f} ms")
print(f"Factor      : {t_bucle/t_vect:8.0f}x más rápido")
print("¿Mismo resultado?", np.allclose(a, b))
````

## 5. matplotlib

Los resultados del apartado 4 salían en líneas de texto, y así no se comparan cuatro resoluciones
ni se mira una imagen. matplotlib es la librería que dibuja: convierte una lista de números en una
gráfica y un array de numpy en una imagen, y la deja debajo de la celda, sin fichero intermedio ni
ningún otro programa por medio. Se instaló en el 1.2, junto a numpy.

Se importa siempre igual, con el `as` del 1.3: el submódulo `pyplot`, con el nombre corto `plt`
que es el que aparece en toda la documentación.

````python
import matplotlib.pyplot as plt
````

Hay dos formas de dibujar con matplotlib, y se mezclan mal. La heredada de MATLAB (`plt.plot`,
`plt.title`) aplica cada orden a «la figura actual», que es una variable que la librería guarda por
dentro: `plt.title` rotula la última figura que se haya tocado, sea cual sea. En un cuaderno, con
las celdas ejecutándose en el orden que se quiera, esa última figura puede ser la de cualquier otra
celda.

Usaremos siempre la otra, que no tiene figura actual. `plt.subplots` devuelve la figura y sus ejes
como objetos, se guardan en variables, y cada orden va dirigida al objeto que se nombra.

```python
fig, ax = plt.subplots()                        # una sola gráfica
fig, (ax1, ax2) = plt.subplots(1, 2)            # dos, lado a lado
ax.plot(valores)                                # una curva
ax.set_ylabel("...");  ax.set_title("...")      # rótulos
```

Las celdas terminan con `plt.show()`. Sin él, si la última línea de la celda es una orden de
dibujo, sale además la línea de texto con la que Python describe el objeto que esa orden ha
devuelto.

El 5.1 dibuja una lista de números, que es lo más sencillo que hace la librería; el 5.2 pasa a
dibujar un array de dos dimensiones, que es como se ve una imagen.

### 5.1 Ejemplo resuelto: una curva

`ax.plot` recibe una lista de números y los reparte por el eje vertical. Si no se le da nada más,
usa la posición dentro de la lista como eje horizontal. Sale una curva, que es una función de una
variable: a cada posición le corresponde un valor.

La celda construye `rampa`, 320 números repartidos de 0 a 255, y la dibuja. `np.linspace(0, 255,
320)` da esos 320 valores en coma flotante y `.astype(np.uint8)` los pasa al tipo de un píxel, que
es el del apartado 4. Los rótulos de los ejes se ponen con `set_xlabel` y `set_ylabel`, y el
título con `set_title`.

````python
rampa = np.linspace(0, 255, 320).astype(np.uint8)

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(rampa)
ax.set_xlabel("columna")
ax.set_ylabel("valor del píxel")
ax.set_title("Una rampa de 0 a 255")
fig.tight_layout()
plt.show()
````

La recta sube de 0 a 255 a lo largo de las 320 posiciones, y el eje horizontal las numera sin que
se lo hayamos pedido: son los índices de la lista.

### 5.2 Ejemplo resuelto: imágenes

Una imagen no es una función de una variable, sino de dos: a cada fila y cada columna les
corresponde un valor, y eso ya no cabe en una curva. Para eso está `ax.imshow`, que se pide igual
que `ax.plot` y reparte los números del array por la superficie del dibujo, pero que antes decide
dos cosas por su cuenta.

La primera es el color. Un array de tres dimensiones lo entiende como RGB y lo pinta tal cual. Uno
de dos dimensiones no trae color que pintar, así que le aplica un mapa que traduce cada número a
un color, y el que trae por defecto no es gris sino una escala de azules y amarillos. Para ver el
gris que hay de verdad hay que pedirlo con `cmap="gray"`.

La segunda es el rango. Con un mapa de color, el número más bajo del array se dibuja con el primer
color del mapa y el más alto con el último, sean cuales sean esos dos números. Una imagen gris
cuyos valores vayan de 29 a 76 se dibuja, por tanto, de negro a blanco, cuando en pantalla tendría
que verse una franja oscura que apenas cambia. `vmin=0, vmax=255` fija el rango al del byte y
quita esa normalización.

Los ejes son cosa aparte: numeran filas y columnas, no significan nada en una imagen y se quitan
con `ax.axis("off")`.

El array que se dibuja es un degradado de 240 filas por 320 columnas construido con la `rampa` del
5.1: el canal rojo crece de 0 a 255 y el azul decrece de 255 a 0, que son las dos curvas del
ejercicio anterior repetidas en todas las filas. La celda lo dibuja cuatro veces: en color, en
gris con el rango fijado, en gris sin fijarlo y sin pedir `cmap`.

````python
degradado = np.zeros((240, 320, 3), dtype=np.uint8)
degradado[:, :, 0] = rampa          # rojo creciente
degradado[:, :, 2] = 255 - rampa    # azul decreciente

gris_degradado = a_gris(degradado)
print(f"El gris va de {gris_degradado.min():.0f} a {gris_degradado.max():.0f}")

fig, axes = plt.subplots(1, 4, figsize=(14, 3.2))
axes[0].imshow(degradado)
axes[0].set_title("Color")
axes[1].imshow(gris_degradado, cmap="gray", vmin=0, vmax=255)
axes[1].set_title('cmap="gray", 0 a 255')
axes[2].imshow(gris_degradado, cmap="gray")
axes[2].set_title('cmap="gray", sin vmin ni vmax')
axes[3].imshow(gris_degradado)
axes[3].set_title("sin cmap")
for ax in axes:
    ax.axis("off")
fig.tight_layout()
plt.show()
````

El segundo dibujo y el tercero son el mismo array con el mismo mapa de color, y no se parecen. En
el tercero matplotlib ha estirado el 29 hasta el negro y el 76 hasta el blanco, así que el
degradado sale con todo el contraste; en el segundo, con `vmin` y `vmax` puestos, sale lo que el
array mide de verdad. El cuarto no es gris de ninguna manera, porque el mapa por defecto es de
colores.

Dibujar sin fijar el rango impide comparar dos imágenes entre sí y juzgar si uno tiene algo
oscuro delante, porque cada una se dibuja con su propia escala.

## 6. ffmpeg: llamar a un programa externo

En el entorno hemos instalado numpy y matplotlib, que son librerías: código Python que el
intérprete lee y carga dentro del proceso al encontrarse un `import`.

**ffmpeg no es una librería.** Es un programa, un único archivo ejecutable que conda deja dentro
de la carpeta del entorno, junto al `python.exe`. Un entorno no guarda solo librerías: guarda
también programas, y quedan disponibles para lo que hagamos desde él. Ese archivo funciona por su
cuenta, sin Python: se puede escribir `ffmpeg` en el terminal y hace su trabajo igual.

Por eso `import ffmpeg` no existe. Un `import` sirve para traerse código escrito para ser leído
por el intérprete, y ahí no hay código que leer: hay un programa ya compilado, que se ejecuta
entero de una vez. La única forma de usarlo es lanzarlo, esperar a que acabe y mirar el fichero
que haya dejado en el disco.

Por lo mismo `pip` no puede instalarlo, porque solo gestiona paquetes de Python. Conda sí, porque
también gestiona programas. En el laboratorio ya está instalado. En vuestro PC hay que instalarlo,
con el entorno activado:

```bash
conda install -c conda-forge ffmpeg
```

El símbolo `!` es otra extensión de IPython: manda el resto de la línea al intérprete de comandos
del sistema, en vez de a `python.exe`.

````python
!ffmpeg -version
````

Si en vez de la versión sale que no encuentra la orden, o no está en ese entorno, cosa que se ve
con `conda list ffmpeg`, o el kernel es de otro.

ffmpeg puede generar vídeo sin fichero de entrada, con el dispositivo virtual `lavfi`. Esta celda
produce dos segundos de una carta de ajuste y los guarda en `demo.mp4`, junto al cuaderno.

````python
!ffmpeg -y -loglevel error -f lavfi -i testsrc2=size=320x240:rate=10:duration=2 demo.mp4

from pathlib import Path
demo = Path("demo.mp4")
print("Creado:", demo.exists(), f"({demo.stat().st_size/1024:.1f} kB)")
````

`-loglevel error` es lo que hace que la celda no imprima nada más: ffmpeg escribe toda su
información por `stderr`, también cuando no hay error.
